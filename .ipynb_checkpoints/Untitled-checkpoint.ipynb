```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
# Wczytujemy nasz plik do głównego obiektu DataFrame
df = pd.read_csv('gym_tracker_export.csv', sep=';', encoding='utf-8-sig')

# Wyświetlamy strukturę danych i wykryte typy zmiennych
print('Struktura zbioru')
df.info()

# Podglądamy pierwsze 5 wierszy, żeby ocenić jakość wczytania
print('Próbka danych')
df.head()
```

```python
# Brutalnie wyciągamy sam wzorzec daty (4 cyfry-2 cyfry-2 cyfry) ignorując ukryte śmieci
df['Data'] = pd.to_datetime(df['Data'].astype(str).str.extract(r'(\d{4}-\d{2}-\d{2})')[0])

df.info()
```

```python
df['Tonaz'] = df['Powtorzenia'] * df['Ciezar_kg']

df.head()
```

```python
df.groupby('Partia_miesniowa')['Tonaz'].sum().sort_values(ascending=False)
```

```python
df_bench = df['Cwiczenie'] == 'Bench Press (Barbell)'
wynik = df[df_bench]
print(wynik)
```

```python
x = wynik['Data']
y = wynik['Szacowany_1RM']
plt.plot(x, y)
plt.title('Szacowany 1RM')
plt.xlabel('Data')
plt.ylabel('1RM')
plt.xticks(rotation=45)
plt.show()
```

```python
# 1. Wyciągamy maksymalny 1RM dla każdego dnia
wynik_max = wynik.groupby('Data')['Szacowany_1RM'].max()

# 2. Wyciągamy osie (po groupby data staje się tzw. indeksem)
x = wynik_max.index
y = wynik_max.values

# 3. Rysujemy wykres z punktami (marker='o')
plt.plot(x, y, marker='o')
plt.title('Szczytowy Szacowany 1RM - Wyciskanie Leżąc')
plt.xlabel('Data')
plt.ylabel('1RM (kg)')
plt.xticks(rotation=45) # Obracamy daty dla czytelności
plt.show()
```

```python
def rysuj_1RM(cwiczenie):
    maska = df['Cwiczenie'] == cwiczenie
    wynik = df[maska]

    # 1. Wyciągamy maksymalny 1RM dla każdego dnia
    wynik_max = wynik.groupby('Data')['Szacowany_1RM'].max()

    # 2. Wyciągamy osie (po groupby data staje się tzw. indeksem)
    x = wynik_max.index
    y = wynik_max.values

    # 3. Rysujemy wykres z punktami (marker='o')
    plt.plot(x, y, marker='o')

    # Tworzymy sztuczną oś X z samymi liczbami (0, 1, 2...)
    x_numeryczne = np.arange(len(x))

    # Algorytm ML: obliczamy współczynniki (1 oznacza linię prostą)
    wspolczynniki = np.polyfit(x_numeryczne, y, 1)
    funkcja_trendu = np.poly1d(wspolczynniki)

    # Rysujemy czerwoną, przerywaną linię trendu na naszym wykresie
    plt.plot(x, funkcja_trendu(x_numeryczne), color='red', linestyle='--')
    
    plt.title(f'Szczytowy Szacowany 1RM - {cwiczenie}')
    plt.xlabel('Data')
    plt.ylabel('1RM (kg)')
    plt.xticks(rotation=45) # Obracamy daty dla czytelności
    plt.show()

    przyszly_trening = len(x) + 4
    prognoza = funkcja_trendu(przyszly_trening)

    print(f'Obecny trend przewiduje, że za miesiąc twój maks wyniesie ok. {prognoza:.1f} kg')
```

```python
rysuj_1RM('Front Squat')
```
