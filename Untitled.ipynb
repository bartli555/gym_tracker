```python
import pandas as pd
# Wczytujemy nasz plik do głównego obiektu DataFrame
df = pd.read_csv('gym_tracker_export.csv', sep=';', encoding='utf-8-sig')

# Wyświetlamy strukturę danych i wykryte typy zmiennych
print('Struktura zbioru')
df.info()

# Podglądamy pierwsze 5 wierszy, żeby ocenić jakość wczytania
print('Próbka danych')
df.head()
```

```python
# Brutalnie wyciągamy sam wzorzec daty (4 cyfry-2 cyfry-2 cyfry) ignorując ukryte śmieci
df['Data'] = pd.to_datetime(df['Data'].astype(str).str.extract(r'(\d{4}-\d{2}-\d{2})')[0])

df.info()
```

```python
df['Tonaz'] = df['Powtorzenia'] * df['Ciezar_kg']

df.head()
```
